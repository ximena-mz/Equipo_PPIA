Ejercicio 1: Se crea una función que solicita la exactitud de un modelo mediante input() y utiliza condiciones if, elif y else para clasificar su rendimiento según el porcentaje obtenido.

Ejercicio 2: Se define una tupla con la configuración base y una función que utiliza una lista y un ciclo for para multiplicar las tasas de aprendizaje por un factor de escala de 2.0.

```python
CONFIGURACION_BASE = (512, 10)

def procesar_tasas_aprendizaje(tasas_iniciales, factor_escala=2.0):

    tasas_escaladas = []

    for tasa in tasas_iniciales:
        tasas_escaladas.append(tasa * factor_escala)

    print(f"Configuración base (Tupla): {CONFIGURACION_BASE}")
    print(f"Tasas de aprendizaje originales: {tasas_iniciales}")
    print(f"Tasas de aprendizaje escaladas: {tasas_escaladas}")

    return tasas_escaladas, CONFIGURACION_BASE


learning_rates = [0.1, 0.01, 0.001, 0.0001]

procesar_tasas_aprendizaje(learning_rates)
```

Ejercicio 3: Se crea una función que simula el entrenamiento de un modelo mediante un ciclo while, calculando una pérdida decreciente hasta alcanzar el umbral de tolerancia o el máximo de épocas.

```python
def simular_entrenamiento():

    try:
        max_epocas = int(input("Ingresa el número máximo de épocas a simular: "))

        if max_epocas <= 0:
            print("Ingresa un número de épocas mayor a cero.")
            return

        umbral_tolerancia = 0.05

        print(f"\nIniciando entrenamiento por un máximo de {max_epocas} épocas...\n")

        epoca = 1

        while epoca <= max_epocas:

            loss_actual = 1.0 / (epoca * 0.4)

            print(f"Época {epoca}/{max_epocas} - Loss simulado: {loss_actual:.4f}")

            if loss_actual < umbral_tolerancia:
                print(f"\n¡Convergencia alcanzada! El modelo convergió en la época {epoca}.")
                break

            epoca += 1

        else:
            print("\nEl entrenamiento finalizó sin alcanzar el umbral óptimo.")

    except ValueError:
        print("Por favor, ingresa un número entero válido para las épocas.")


simular_entrenamiento()
```
